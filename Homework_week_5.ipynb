```python
df["year_num"] = pd.to_numeric(df["year"].astype(str).str[:4], errors="coerce")
df["runtime_min"] = df["runtime"].astype(str).str.extract(r"(\d+)")[0].astype(float)
df["votes"] = pd.to_numeric(df["imdb_votes"].astype(str).str.replace(",", ""), errors="coerce")
df[["title", "year_num", "runtime_min", "votes"]].head()
```

```python
df.isna().sum()
```

```python
df = df.dropna(subset=["title"])
df.isna().sum()
```

```python
movies = df[df["type"] == "movie"]
movies[(movies["year_num"] >= 2010) & (movies["imdb_rating"] > 7.5)][["title", "year_num", "imdb_rating"]]
```
